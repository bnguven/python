````python
def rectangle_area(width, height):
    """
    Calculates the area of a rectangle

    example: rectangle_area(3,4) # 12
    """
    return(width * height)
````

````python
def larger_number(a, b):
    """
    Returns the larger of two numbers

    example: larger_number(7, 8) # 8

    """
    if a > b:
        return a
    else:
        return b

larger_number(7, 7860)
````
